```python
import csv
from datetime import datetime
```

```python
FILENAME = 'expenses.csv'
```

```python
#load expenses
def load_expenses():
    expenses = []
    try: 
        with open(FILENAME, 'r', newline ='') as file:
            reader = csv.DictReader(file)
            for row in reader:
                row['amount'] = float(row['amount'])
                expenses.append(row)

    except FileNotFoundError:
        pass
    return expenses

```

```python
#saving expenses
def save_expenses(expenses):
    with open(FILENAME, mode = 'w', newline = '') as file:
        fieldnames = ['date', 'category', 'amount', 'description']
        writer = csv.DictWriter(file, fieldnames= fieldnames)
        writer.writeheader()
        for expense in expenses:
            writer.writerow(expense)
        
```

```python
#adding a new expense
def add_expense(expenses):
    date_str = input("enter date (YYYY-MM-DD): ")
    try:
        datetime.strptime(date_str, '%Y-%m-%d')
    except ValueError:
        print("invalid date format.")
        return 
    category = input("Enter category: ")
    try:
        amount = float(input("Enter amount: "))
    except ValueError:
        print("Invalid amount.")
        return
    description = input("enter the description: ")
    expenses.append({
        'date': date_str,
        'category': category,
        'amount': amount,
        'description': description })
    print("Expense added!")
```

```python
#vewing all the expenses

def view_the_expense(expenses):
    if not expenses:
        print("No Expenses found")
        return

    print(f"{'Index':<6}{'Date':<12}{'Category':<15}{'Amount':<10} Description")
    print("-" * 60)
    for i, exp in enumerate(expenses):
        print(f"{i:<6}{exp['date']:<12} {exp['category']:<15} {exp['amount']:<10.2f} {exp['description']}")
```

```python
#deleting an expense by index

def delete_expenses(expenses):
    view_expenses(expenses)
    try:
        idx = int(input("enter the index of the expense to delete: "))
        if 0<=idx<len(expenses):
            removed = expenses.pop(idx)
            print(f"Removed expense : {removed}")
        else:
            print("Invalid index.")

    except ValueError:
        print("invalid input.")
```

```python
#main loop with menu
def main():
    expenses = load_expenses()
    while True:
        print("\nPersonal expense tracker")
        print("1. Add Expenses")
        print("2. View expenses")
        print("3. Delete Expenses")
        print("4. Save and Exit")
        choice = input("Enter your choice: ")

        if choice =='1':
            add_expense(expenses)
        elif choice =='2':
            view_expenses(expenses)
        elif choice =='3':
            delete_expense(expenses)
        elif choice =='4':
            save_expenses(expenses)
            print("Expenses saved.")
            break
        else:
            print("Invalid input, Please try again")

if __name__=='__main__':
    main()
```
